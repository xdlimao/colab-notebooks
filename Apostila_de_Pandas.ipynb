## Nível 8: Operações Avançadas (Join, Pivot, Apply)

```python
# 36 - Transforme a coluna 'ID_Venda' no índice oficial do DataFrame (set_index). (Resposta: A primeira coluna exibida à esquerda será o ID, começando do 1)
df_36 = df.copy()
df_36.set_index("ID_Venda")
```

```python
# 37 - Crie este pequeno DataFrame rodando este código primeiro:
df_impostos = pd.DataFrame({'Categoria': ['ELETRÔNICOS', 'COMPLEMENTOS'], 'Imposto': [0.15, 0.05]})
# Agora faça um merge (join) do seu DataFrame original (que agora deve ter COMPLEMENTOS no lugar de Acessórios) com esse df_impostos usando a coluna 'Categoria'. (Resposta: O DataFrame ganhará uma coluna 'Imposto')
df_37 = df.copy()

df_37["Categoria"] = df_37["Categoria"].str.replace("Acessórios", "Complementos")
df_37["Categoria"] = df_37["Categoria"].str.upper()
df_37.merge(df_impostos)
df_37
```

```python
# 38 - Usando a função apply e uma função lambda, crie uma coluna chamada 'Ticket_Medio_Item' que seja o 'Valor_Bruto' dividido pela 'Quantidade'. (Resposta: Nova coluna, o primeiro valor será 3500.0)
df_38 = df.copy()
df_38["Valor_Bruto"] = df_38["Quantidade"] * df_38["Preco_Unitario"]
calculo_tmi = lambda x: (x["Valor_Bruto"]/x["Quantidade"])
df_38["Ticket_Medio_Item"] = df_38.apply(calculo_tmi, axis=1)
df_38
```

```python
# 39 - Crie uma Tabela Dinâmica (pivot_table) onde os Índices (linhas) sejam os 'Produtos', as Colunas sejam as 'Categorias' e os Valores sejam a soma da 'Quantidade'. Preencha os valores nulos com zero (fill_value=0). (Resposta: Matriz mostrando quantidades vendidas cruzando produto e categoria).
df_39 = df.copy()
df_39["Produto"] = df_39["Produto"].fillna("Produto Desconhecido") # Preenche NaN para que não sejam ignorados no pivot_table
pivot_table_resultado = pd.pivot_table(df_39, values='Quantidade', index='Produto', columns='Categoria', aggfunc='sum', fill_value=0)
display(pivot_table_resultado)
```

```python
# 40 - Ordene o DataFrame original pela coluna 'Data' de forma crescente. Em seguida, crie uma coluna 'Faturamento_Acumulado' que calcula a soma acumulada (cumsum()) da coluna 'Valor_Final' linha a linha. (Resposta: A última linha dessa coluna representará o faturamento total da loja no período todo: 15842.5).
df_40 = df.copy()

# Certifica que a coluna 'Data' é do tipo datetime para ordenação correta
df_40["Data"] = pd.to_datetime(df_40["Data"])

# Calcula 'Valor_Bruto' e 'Valor_Final' se ainda não existirem
df_40["Valor_Bruto"] = df_40["Quantidade"] * df_40["Preco_Unitario"]
df_40["Valor_Final"] = df_40["Valor_Bruto"] * (1 - df_40["Desconto"])

# Ordena o DataFrame pela coluna 'Data'
df_40_ordenado = df_40.sort_values(by='Data', ascending=True)

# Calcula a soma acumulada do 'Valor_Final'
df_40_ordenado["Faturamento_Acumulado"] = df_40_ordenado["Valor_Final"].cumsum()
display(df_40_ordenado)
```

## Nível 14: Preenchendo Buracos (fillna)

```python
# 53 - No seu df_limpo, você notará que o Bruno e o Eduardo (índices 1 e 4) não têm idade. Em vez de apagar esses clientes, preencha os valores vazios da coluna 'Idade' com a média das idades válidas que restaram no df_limpo. Salve a alteração no próprio df_limpo. (Resposta: A idade do Bruno e do Eduardo passará a ser 37.25).
df_limpo["Idade"] = df_limpo["Idade"].fillna(df_limpo["Idade"].mean())
```

```python
# 54 - Ainda no df_limpo, preencha os valores faltantes de duas colunas com lógicas diferentes: tudo que for nulo em 'Valor_Gasto' deve virar 0.0 (pois a pessoa não comprou nada), e o que for nulo em 'Cidade' deve virar a string 'Não Informada'. (Resposta: A Fernanda, no índice 5, terá Valor_Gasto de 0.0. O Gabriel, no índice 6, terá a Cidade como 'Não Informada').
# Nota: Devido à remoção de linhas na questão 52, 'Fernanda' e 'Gabriel' podem não estar presentes nos índices mencionados, mas a lógica de preenchimento será aplicada aos valores nulos restantes.
df_limpo["Valor_Gasto"] = df_limpo["Valor_Gasto"].fillna(0.0)
df_limpo["Cidade"] = df_limpo["Cidade"].fillna("Não informado")
df_limpo
```
