### Importando os dados com Pandas

````python
import pandas as pd
#Importando o nosso dataset para manipularmos gráficos!
df = pd.read_csv("https://raw.githubusercontent.com/xdlimao/colab-notebooks/refs/heads/main/arquivos_csv/imigracao_por_paises.csv")
df.set_index("País", inplace=True)
df
````

````python
df.info()
````

````python
#Vamos organizar os dados para Anos e Imigrantes
anos = list(map(str, range(1980, 2014)))
brasil = df.loc["Brasil", anos]

brasil_dict = {
    "anos": brasil.index.tolist(),
    "imigrantes": brasil.values.tolist()
}

#Criando o DF com Pandas
dados_brasil = pd.DataFrame(brasil_dict)
dados_brasil
````

### Nosso primeiro gráfico!

````python
# Nós chamamos o módulo pyplot, que seria o módulo referente na criação dos gráficos
# E convencionalmente definimo ele como plt
import matplotlib.pyplot as plt

"""
  plt.figure(figsize=(eixo_x, eixo_y))
  Esse método indica qual o tamanho do gráfico em si, estando em polegadas
  Deve ser chamado antes do plt.plot
"""
plt.figure(figsize=(20, 4))


"""
  Aqui está o principal:
  plt.plot(eixo_x, eixo_y)
  Com ele, você cria um gráfico de linhas
"""
plt.plot(dados_brasil["anos"], dados_brasil["imigrantes"])



# Com as funções abaixo, conseguimos colocar título do gráfico, nome do eixo x e nome do eixo y
plt.title("Imigração do Brasil para o Canada")
plt.xlabel("Ano")
plt.ylabel("Número de imigrantes")
plt.xticks(rotation=45) #E até rotacionar as labels xd

"""
  Ticks se refere ao tracinho pequeno que existe nas retas.
  Caso você gere valores fora do dados, ira deixar uma área em branco

  Para colocar os ticks de forma manual, utilize as funções
  abaixo e passe os arrays
"""
plt.xticks([str(x) for x in range(1980, 2011, 5)])
plt.yticks([x for x in range(500, 3001, 500)])

# Aqui mostramos o gráfico para olharmos
plt.show()
````

### Como mostrar gráficos no output
## 📊 Exibição de gráficos no Matplotlib (Notebooks)

Métodos comuns:

* `plt.show()`
* `fig.show()`
* `display(fig)` (de IPython.display)
* `fig` → objeto retornado por `plt.subplots()`

---

## 🧠 Comparação resumida

| Método                                    | O que mostra               | Quando mostra   | Entre células?                  | Ideia principal                  |
| ----------------------------------------- | -------------------------- | --------------- | ------------------------------- | -------------------------------- |
| `plt.show()`                              | Todas as figuras pendentes | Na execução     | ❌ Não                           | Mostra e encerra                 |
| `fig.show()`                              | Apenas aquela figura       | Na execução     | ❌ Não (pode variar por backend) | Controle direto da figura        |
| `display(fig)`                            | Apenas aquela figura       | Quando chamado  | ✅ Sim                           | Flexível, funciona entre células |
| `fig` (última linha, de `plt.subplots()`) | Apenas aquela figura       | Final da célula | ❌ Não                           | Renderização automática          |

---

## 🔍 Exemplos compactos

```python
import matplotlib.pyplot as plt
from IPython.display import display

# Criação moderna de figura e eixo
fig, ax = plt.subplots()  # 'fig' é a figura, 'ax' é o eixo
ax.plot([1,2,3])

# --- Na mesma célula ---
plt.show()      # mostra e “fecha” tudo
fig.show()      # mostra só esta figura (pode variar)
display(fig)    # mostra só esta figura (flexível)
fig             # última linha: renderiza automaticamente
```

```python
# --- Em outra célula ---
plt.show()      # ❌ não mostra o gráfico anterior
fig.show()      # ❌ geralmente não mostra em notebooks (no Colab funcionou)
display(fig)    # ✅ funciona porque 'fig' ainda existe
```

---

## 🔑 Resumo mental

* `plt.show()` → mostrar **todas** as figuras agora e encerrar
* `fig.show()` → mostrar **uma figura** agora (menos confiável em notebooks)
* `display(fig)` → mostrar **uma figura** quando quiser, até em outra célula
* `fig` → objeto retornado por `plt.subplots()`, usado para controle direto da figura

## 💡Sobre o plt.show()
Ele só mostra **as figuras que estão “pendentes” no estado global do matplotlib no momento da execução**.

Em notebooks, **cada célula executa separadamente**, então figuras criadas em uma célula anterior não estão mais pendentes para `plt.show()` na célula seguinte.

💡 Resumindo: **plt.show() não “lembra” gráficos de células anteriores**.

---

## ✅ Dicas práticas

* Notebook simples → última linha `fig` ou `display(fig)`
* Controle e atualização → `display(fig)`
* Código portável (script + notebook) → `plt.show()`

##Criando figuras com Matplotlib

###Um resumo do que você ira ver

Se você já se sentiu meio perdido com o `fig, ax = plt.subplots()`, não se preocupe: essa é a "porta de entrada" para o modo profissional de usar o Matplotlib (chamado de **Interface Orientada a Objetos**).

Basicamente, essa linha decompõe o seu gráfico em duas partes fundamentais: a **moldura** e o **conteúdo**.

---

### O que cada um faz?

#### 1. O `fig` (Figure)
A **Figure** é o objeto de nível superior. Pense nela como a **folha de papel** ou a janela inteira do gráfico. Ela gerencia tudo o que está nela, mas não "desenha" os dados diretamente.
* **Para que serve:** Alterar o tamanho da imagem (`figsize`), salvar o arquivo (`savefig`) ou configurar o layout geral.

#### 2. O `ax` (Axes)
O **Axes** é o gráfico em si — a área onde os dados são plotados. É aqui que você define o que vai nos eixos $x$ e $y$.
* **Para que serve:** Desenhar linhas, barras, definir títulos, nomes dos eixos e legendas.



---

### Por que não usar apenas `plt.plot()`?

Você provavelmente já viu exemplos mais simples usando apenas `plt.plot()`. A diferença é que o `fig, ax` te dá **controle total**, especialmente quando você quer criar vários gráficos na mesma imagem.

| Método | Analogia | Uso Ideal |
| :--- | :--- | :--- |
| **pyplot (plt.plot)** | Câmera automática: ela decide quase tudo por você. | Gráficos rápidos e simples de uma única tela. |
| **Object-Oriented (fig, ax)** | Câmera manual: você ajusta cada detalhe individualmente. | Projetos complexos, múltiplos subgráficos ou customização fina. |

---

### Exemplo Prático: Múltiplos Gráficos
A maior vantagem aparece quando você quer, por exemplo, dois gráficos lado a lado:

```python
import matplotlib.pyplot as plt

# Criamos 1 linha e 2 colunas de gráficos
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# Acessamos cada gráfico pelo índice do array 'ax'
ax[0].plot([1, 2, 3], [10, 20, 10], color='red')
ax[0].set_title("Gráfico da Esquerda")

ax[1].bar(['A', 'B', 'C'], [5, 7, 3])
ax[1].set_title("Gráfico da Direita")

plt.show()
```

---

> **Dica de ouro:** Sempre que vir um método que começa com `set_` (como `ax.set_xlabel()`), você está usando a interface orientada a objetos. No `plt`, o comando seria apenas `plt.xlabel()`.

````python
#Criando a manipulação de figura e eixos
fig, ax = plt.subplots(
    figsize=(8, 4) # Já conseguimos passar o tamanho do gráfico direto na criação
)

"""
  Basicamente criamos a forma: fig e ax

  Agora vamos usar de fato!
  Vamos criar uma gráfico de linha na forma mais avançada do matplotlib
"""
ax.plot(dados_brasil["anos"], dados_brasil["imigrantes"])

"""
  Lembra da dica do texto (que resumiu muita coisa xD)?
  set_..., então se eu quero plt.title, seria ax.set_title!
"""
ax.set_title("Imigração no Brasil para o Canadá\n1980 a 2013")
ax.set_xlabel("Anos")
ax.set_ylabel("Número de imigrantes")

"""
  Ao invez de plt.xticks para definirmos os ticks manualmente, vamos fazer
  os valores serem um padrão (que seria uma sequência de multiplicador) de forma dinâmica
  (Adendo: ax.set_xticks existe, entendeu a lógica?)
"""
ax.xaxis.set_major_locator(plt.MultipleLocator(5))

plt.show()
````

###Multiplos gráficos em forma de vetor



````python
"""
  Olha que interessante, temos como criar uma tabela com vários gráficos!
  Queremos criar uma visualização assim

  Tabelas (1, 2)
  XX XX

  Se quiser criar uma visualização de 2 linhas e 3 colunas, ficaria assim
  Tabelas(2, 3)
  XX XX XX
  XX XX XX
"""
fig, axs = plt.subplots(
    1, #Quantidade de linhas = Eixo X. Padrão 1
    2, #Quantidade de colunas = Eixo Y. Padrão Y
    figsize=(15, 5)
)

"""
  Como criamos um vetor de tabelas, usamos apenas 1 colchete para selecionar a coluna
"""
axs[0].plot(dados_brasil["anos"], dados_brasil["imigrantes"]) #Cria gráfico
axs[0].set_title("Imigração do Brasil para o Canadá\n1980 a 2013") #Título
axs[0].set_xlabel("Ano") #Legenda no eixo X
axs[0].set_ylabel("Número de imigrantes") #Legenda no eixo Y
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5)) #Multiplicador dos dados no eixo X

"""
  Algo que não vimos até agora, da para adicionar linhas para
  ajudar na visualização. Obrigado Grids por existirem xD
"""
axs[0].grid()

#E repete mais do mesmo em baixo...
axs[1].boxplot(dados_brasil["imigrantes"])
axs[1].set_title("Boxplot da imigração do Brasil para o Canadá\n1980 a 2013")
axs[1].set_xlabel("Brasil")
axs[1].set_ylabel("Número de imigrantes")
axs[1].grid()

"""
Como estamos começando a trabalhar com gráficos com várias informações. Uma coisa pode estar em cima da outra
atrapalhando a visualização. Para isso temos o tight layout que ajusta as margens e tamanhos automáticamente.
Gráficos muitos complexos com vários de uma vez nem sempre se resolvem sozinho, mesmo com o plt.tight_layout()
-> Normalmente se ve em matrizes
"""
#plt.tight_layout()

plt.show()
````

###Criando multiplos gráficos em forma de matriz

````python
"""
  Opa, parece que vai ser 2 por 2

  Tabelas
  XX   XX

  XX   XX
"""
fig, axs = plt.subplots(
    2,
    2,
    figsize=(15, 10) #Agora é importante deixar bem grande para ter espaço para todos (ou o matplotlib se ajusta)
)

"""
  Aqui temos uma função em que da o espaçamento entre os gráficos, seria o nosso
  equivaliente do margin do css. Valores entre 0.2 e 1 já estão ok!
"""
fig.subplots_adjust(hspace=0.5, wspace=0.3)

# E aqui nós colocamos um título geral para os 4 gráficos existentes
fig.suptitle("Imigração dos quatros maiores países da Ámerica do Sul para o Canadá")

axs[0, 0].plot(df.loc["Brasil", anos])
axs[0, 0].set_title("Brasil")

axs[0, 1].plot(df.loc["Colômbia", anos])
axs[0, 1].set_title("Colômbia")

axs[1, 0].plot(df.loc["Argentina", anos])
axs[1, 0].set_title("Argentina")

axs[1, 1].plot(df.loc["Peru", anos])
axs[1, 1].set_title("Peru")

"""
  Conseguimos também manipular os gráficos dentro de alguns loops que podemos fazer lógica para
  manter padrões, alterações dinamicas e compartilhar valores de variáveis!

  Vamos fazer duas lógicas simples para você pegar o jeito:
  1 -> Vamos definir que o multiplicador de dados seja a cada 5 casas de diferença, colocar suas
  'label x' & 'label y' e por último ativar o grid de todas as tabelas

  2 -> Vamos deixar todos os limites de gráficos seguindo o mesmo valor, assim temos uma ideia
  e visualização melhorada pois todos os países estão no mesmo contexto.
"""

#Valores para deixar todos os gráficos iguais
ymin = 0
ymax = 7000

"""
  O axs.flat retorna um iterador que passa por cada eixo (que nesse contexto o correto é gráfico, mas enfim)
  em que você só passa ele no for e pronto. Já pode manipular os dois ao mesmo tempo
"""
for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.set_xlabel("Ano")
  ax.set_ylabel("Número de imigrantes")
  ax.grid()

"""
  Agora, o ravel() é uma função do numpy que transforma matriz em um vetor iterador.
  Deixou de ser [[..., ...], [..., ...]] => [..., ..., ..., ...]
  Por que utilizamos ele aqui?
  Apenas para mostrar mais forma de usar outros tipos de iteradores

  Os dois tem o mesmo funcionamento nesse contexto de iterar
"""
for ax in axs.ravel():
  ax.set_ylim(ymin, ymax)

"""
A regra de bolso é:
Use .flat se você quer apenas fazer um loop (é mais "limpo").
Use .ravel() se você precisar fazer algo além do loop, como selecionar apenas os 3 primeiros eixos: axs.ravel()[:3]
"""

#Olha o tight_layout aqui de novo. No caso da pra ver que temos muito espaço no gráfico. Se for rodar o tight_layout
#ele vai reduzir aqui nesse caso os espaços pois ele só quer o minimo possível para nenhum elemento ficar em cima do outro.
#plt.tight_layout()
plt.show()
````

##Customizando com Matplotlib

####Básico

Referência rápida, não precisa olhar agora, só se quiser saber alguns valores de customização:

---
Cores:
|Símbolo |	Resultado |	Descrição|
|-|-|-|
|.|Ponto	|Ponto pequeno|
|o|Círculo|Círculo preenchido (o mais comum)|
|s|Square|Quadrado|
|D|Diamond|Diamante|
|^|Triângulo|Triângulo para cima|
|v|Triângulo|Triângulo para baixo|
|x|X|Um "x"|
|+|Plus|Um sinal de mais|
|*|Estrela|Uma estrela de 5 pontas|
|h|Hexagon|Hexágono|

---
Tracejado:
|Comando (ls=)|Símbolo|Resultado Visual|
|-|-|-|
|'solid'|'-'|Linha contínua (padrão)|
|'dashed'|'--'|Linha tracejada (traços longos)|
|'dotted'|':'|Linha pontilhada|
|'dashdot'|'-.'|Traço e ponto intercalados|

````python
# Criando nosso gráficos na forma avançada
fig, ax = plt.subplots(figsize=(8, 4))

# E vamos plotar os dados!
ax.plot(
    dados_brasil["anos"],
    dados_brasil["imigrantes"],
    lw=3,      #Opa, temos como adicionar estilização nos gráficos
    marker="o" #E elementos como pontos na partes em que o dado aparece
)

"""
  Veja alguns elementos que pode ser adicionado durante o gráfico (específico do plot):
  lw=3,                    # Grossura da Linha
  ls='--',                 # Estilo da linha
  marker='o',              # Marcadores para os dados
  markersize=10,           # Tamanho do marcador
  markerfacecolor='white', # Cor interna do marcador
  markeredgecolor='red'    # Cor da borda do marcador
"""

"""
  Vamos manipular agora o texto:
"""
ax.set_title(
    "Imigração no Brasil para o Canadá\n1980 a 2013",
    fontsize=20, #Também conseguimos controlar o tamanho da fonte de um texto!
    loc="left"   #loc = localização, Onde o (título) ficará na visualização. Temos 'left', 'center' e 'right'
)

ax.set_xlabel("Anos", fontsize=12) #Mesma coisa do fontsize de cima
ax.set_ylabel("Número de imigrantes", fontsize=12) #Aqui também

"""
  Para manipular os labels, que seriam os textos dos dados
  É necessário selecionar o eixo, que seria 'xaxis' ou 'yaxis'.
  Após isso, um leque de opções estão disponíveis para nós podermos mexer.

  Para saber de mais valores para manipular, vide doc.
"""
ax.xaxis.set_tick_params(
    labelsize=20 #Aqui é apenas o tamanho da fonte da label
)
ax.yaxis.set_tick_params(labelsize=8)

ax.xaxis.set_major_locator(plt.MultipleLocator(5))

plt.grid(linestyle="--") #Lembra do ls lá de cima? Aqui tem que ser o nome inteiro para alterar o grid
#Vide doc para alterar o grid

plt.show()
````

####Estilos
Além disso que vimos acima, segundo a documentação da biblioteca Matplotlib podemos aplicar diferentes estilos para permitir que nós adaptemos as visualizações de acordo com as nossas necessidades. Para saber quais estilos estão disponíveis, podemos imprimir style.available, onde style é um submódulo da biblioteca Matplotlib utilizado para definir estilos de visualização:

````python
for style in plt.style.available:
  print(style)
````

Antes de aplicar um estilo, é importante saber que cada vez que a biblioteca Matplotlib é importada em um notebook, ela define uma configuração de tempo de execução que inclui os estilos padrão para cada elemento de plotagem criado.

Portanto, para evitar que o estilo seja aplicado a todos os gráficos plotados no mesmo notebook, podemos utilizar um código que cria uma cópia das configurações padrão de plotagem da biblioteca Matplotlib e as atribui à variável IPython_default. Isso pode ser útil para armazenar e reutilizar as configurações padrão de plotagem ou para restaurá-las depois de terem sido modificadas:




````python
IPython_default = plt.rcParams.copy() #Salvamos o estilo padrão do matplotlib

"""
Um dos estilos disponíveis é baseado em um site de notícias e análises de dados chamado
FiveThirtyEight, que cobre assuntos como política, economia, cultura, ciência e esportes.

Para utilizar o estilo 'fivethirtyeight' usamos o código abaixo
"""

plt.style.use('fivethirtyeight')
````

````python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(dados_brasil['anos'], dados_brasil['imigrantes'])
ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', loc='left')
ax.set_ylabel('Número de imigrantes')
ax.set_xlabel('Ano')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.show()
````

Bom, mas e se a gente quiser plotar gráficos sem esse estilo depois de ter definido ele no notebook? Podemos redefinir os parâmetros utilizando rcParams.update e passando a ele a variável IPython_default que criamos anteriormente com as configurações padrão:

````python
plt.rcParams.update(IPython_default);
````

Além disso, podemos aplicar um estilo apenas a um bloco de código específico, pois o pacote de estilo fornece um gerenciador de contexto para limitar suas alterações a um escopo específico. Para isolar suas alterações de estilo, você pode escrever o código dentro de um contexto with da seguinte forma:

````python
with plt.style.context('fivethirtyeight'): # <- Isola o gráfico, também funciona para vetores e matrizes
  fig, ax = plt.subplots(figsize=(8, 4))
  ax.plot(dados_brasil['anos'], dados_brasil['imigrantes'])
  ax.set_title('Imigração do Brasil para o Canadá\n1980 a 2013', loc='left')
  ax.set_ylabel('Número de imigrantes')
  ax.set_xlabel('Ano')
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  plt.show()
````

````python
fig, ax = plt.subplots(figsize=(17, 5))

"""
  Agora, se a gente quer destacar um valor para visualizar, se liga na lógica abaixo.
  Lembrando que definimos o indexes para ser o nome dos países lá no começo desse Notebook:
"""
cores = []
for pais in america_sul.index:
    if pais == 'Brasil':
        cores.append('green')
    else:
        cores.append('silver')

# O bar e o barh é a mesma coisa, exceto que o 'h' indica que é na horizontal
ax.barh(america_sul.index, america_sul["Total"], color=cores) #Colocado as cores na ordem, veja como fica a visualização.

#Mais do mesmo
ax.set_title("Imigração da América do Sul para o Canadá\n1980 a 2013", loc="left", fontsize=18)
ax.set_ylabel("País", fontsize=14)
ax.set_xlabel("Número de imigrantes", fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)
plt.show()

#Fica fácil de achar o Brasil
````

````python
#Mais do mesmo para termos um gráfico de barras na horizontal
fig, ax = plt.subplots(figsize=(17, 5))
ax.barh(america_sul.index, america_sul["Total"])
ax.set_title("Imigração da América do Sul para o Canadá\n1980 a 2013", loc="left", fontsize=18)
ax.set_xlabel("Número de imigrantes", fontsize=14)
ax.yaxis.set_tick_params(labelsize=12)
ax.xaxis.set_tick_params(labelsize=12)

"""
  Vamos supor que só o .grid() não ajude totalmente porque temos pessoas com visão horrível.
  Nesses casos, que tal colocar uma anotação dentro do gráfico? Nice.

  Legenda de uso do for:
  i = indíce (que seria cada barra)
  v = valor (de cada barra)
  enumarate(df[...]) = Quando você faz, por exemplo, enumerate(america_sul["Total"]), o Python transforma
                       aquela coluna de números em pares de (posição, valor). Sem o enumerate, ele le
                       apenas o 'v' e não sabe qual barra mexer.

  Agora explicado o for, vamos entender o .text(...):
"""
for i, v in enumerate(america_sul["Total"]):
  ax.text(
      # Posição de inserção + Texto a inserir

      # Basicamente, o eixo_x e eixo_y segue a coordenada do gráfico mesmo.

      # Se nós colocarmos 0 no eixo_x, a anotação vai aparecer no início da
      # barra (no nosso exemplo, do lado direito do país).

      # Existe uma lógica boa que seria, caso o valor seria a posição X do meu eixo, eu posso passar
      # ela para ficar sempre na ponta da barra (se for no contexto de barra xD), Veja o código:

      v+20,   #eixo_x. Lembrando que o for ira retornar 1, 2, 3... Mas poderia ser 0.1, 0.2, ...
      i,      #eixo_y. Mesmo comentário acima sobre 0.1, 0.2, ...
      str(v), #Valor a ser inserido. Sempre string

      # Estilização
      color="black",
      fontsize=10,
      ha="left",  #Alinhamento horizontal. O texto vai na coordenada, mas ele pode ficar no 'left', 'center' ou 'right' que nem no word.
      va="center" #Alinhamento vertical. Mesma coisa do de cima.
  )

 # Como eu anotei coisa demais, aqui o comentário com a função limpa xD
 # ax.text(v+20, i, str(v), color="black", fontsize=10, ha="left",va="center")


"""
  Boa, agora sobre os desenhos(e réguas) do gráfico:
  .set_frame_on(bool) = Moldura, sendo True o padrão que é mostrar e False para esconder.

  .get_xaxis() -> Opções de manipulação do eixo X (mas poderia ser get_yaxis)
    .set_visible(bool) = Define se mostra a linha do eixo escolhido, sendo True o padrão que é mostrar e False para esconder

  .tick_params(axis=..., which=..., length=...) = Altera os ticks que são realmente os gravetos que fican na linha do eixo
  Veja o bloco de texto abaixo para ter uma visão e explicação mais bonita ao invez daqui dentro dos comentários
"""
ax.set_frame_on(False)
ax.get_xaxis().set_visible(True)
ax.tick_params(axis="both", which="both", length=0) #Altera os ticks do eixo Y, fazendo eles sumirem as pontinhas que começa na barra

plt.show()
````

Essa linha de código serve para **esconder ou personalizar os "tracinhos"** (chamados de *ticks*) que ficam nos eixos $x$ e $y$ do gráfico.

Sabe aqueles pequenos traços que apontam para os números 0, 10, 20 nos eixos? O `tick_params` é o comando que controla a aparência deles.

---

***O que cada parte faz:***

1. `axis="both"`
* **O que define:** Em qual eixo você quer mexer.
* **"both"**: Aplica as mudanças tanto no eixo **Horizontal (X)** quanto no **Vertical (Y)** ao mesmo tempo.
* *Alternativas:* Você poderia usar `axis="x"` ou `axis="y"`.

2. `which="both"`
* **O que define:** Quais tipos de tracinhos serão afetados.
* **"both"**: Afeta os traços **Principais** (*Major*, que seria 'major') e os **Secundários** (*Minor*, que seria 'minor').
* *Contexto:* O Matplotlib geralmente mostra apenas os Major (os que têm números). Se você habilitou divisões menores entre os números (gravetinhos que ficam entre os principais), o "both" garante que todos sumam.

3. `length=0`
* **O que define:** O comprimento (tamanho) do tracinho.
* **`0`**: Ao definir o tamanho como zero, você torna o tracinho **invisível**.
* *Resultado:* Os números continuam lá, mas aquele "dentinho" de marcação desaparece.

---

***Por que usar isso?***<br>
Geralmente usamos `length=0` para deixar o gráfico com um visual mais **limpo (Clean Design)**.

* **Com grades:** Se você já usa `ax.grid()`, os tracinhos nos eixos costumam ser redundantes e "poluem" a imagem.
* **Gráficos de barras:** Em gráficos de barras horizontais (como o seu de imigrantes), é muito comum remover esses traços para focar apenas na barra e no nome do país.

---

***Resumo da ópera:***
> "Vá em **todos** os eixos, pegue **todos** os tipos de tracinhos e deixe o **tamanho** deles em **zero** (esconda-os)."

Depois disso, vamos criar um bloco de código onde vamos construir essa animação, importando `plotly.graph_objs`, um módulo da biblioteca Plotly que contém classes para criar visualizações de dados interativas e personalizadas.

```python
import plotly.graph_objs as go
```

### 3 - Criação da figura
Em seguida, uma figura vazia é criada usando a função `go.Figure()` e atribuída à variável `fig`.

```python
fig = go.Figure()
```

### 4 - Adição do traço inicial
A seguir, uma linha é adicionada ao gráfico usando a função `fig.add_trace()`. Nesta função, é passado um objeto `Scatter`, que recebe como argumentos os dados para os eixos X e Y do gráfico. Para que o gráfico seja exibido sem linha antes de clicar no botão play usamos o `iloc[0]` nas duas variáveis. Isso ocorre porque o `iloc[0]` seleciona o primeiro valor das colunas ano e imigrantes dos dados do Brasil, respectivamente. Ao adicionar esse ponto de dados à visualização do gráfico, ele será exibido inicialmente como um único ponto, sem linhas que o conectem a outros pontos. Além disso, passamos o modo de exibição `lines`, que quer dizer linhas e o nome da linha. Também é definida a espessura da linha usando o dicionário `line=dict(width=4)`.

```python
fig.add_trace(
    go.Scatter(x=[dados_brasil['ano'].iloc[0]], y=[dados_brasil['imigrantes'].iloc[0]], mode='lines', name='Imigrantes', line=dict(width=4))
)
```

### 5 - Configuração do layout
Depois disso, o título do gráfico e as configurações do eixo X e Y são definidas usando a função `fig.update_layout()`. Os argumentos do título são:

*   `text='<b>Imigração do Brasil para o Canadá no período de 1980 a 2013</b>'`: define o texto do título como uma string formatada em negrito (usando as tags HTML "`<b>`" e "`</b>`").
*   `x=0.12`: define a posição horizontal do título no layout, em relação à largura da figura. O valor 0.12 especifica que o título começará a 12% da largura da figura.
*   `xanchor='left'`: define o alinhamento horizontal do título. O valor 'left' significa que o título será alinhado à esquerda do layout.
*   `font=dict(size=20)`: define o tamanho do texto do título.

Já os argumentos para `xaxis` e `yaxis` são dicionários, com as seguintes propriedades:

*   `range=[1980, 2013]`: define o intervalo do eixo, ou seja, o menor e o maior valor que serão exibidos. Neste caso, o eixo x terá como menor valor 1980 e como maior valor 2013, enquanto o eixo y terá como menor valor 0 e como maior valor 3000.
*   `autorange=False`: define se os limites do eixo serão ajustados automaticamente (True) ou não (False). Neste caso, os limites não serão ajustados automaticamente.
*   `title='<b>Ano</b>'`: define o título do eixo. Neste caso, o eixo x terá o título "Ano", que é formatado em negrito.
*   `title='<b>Número de imigrantes</b>'`: define o título do eixo. Neste caso, o eixo y terá o título "Número de imigrantes", que é formatado em negrito.

```python
fig.update_layout(
    title=dict(
        text='<b>Imigração do Brasil para o Canadá no período de 1980 a 2013</b>',
        x=0.12,
        xanchor='left',
        font=dict(size=20)
    ),
    xaxis=dict(range=[1980, 2013], autorange=False, title='<b>Ano</b>'),
    yaxis=dict(range=[0, 3000], autorange=False, title='<b>Número de imigrantes</b>'),
```

### 6 - Adição do botão "Play"
É adicionado um botão "Play" para a animação usando o argumento `updatemenus`. Esse argumento é uma lista que define as opções de menu para a figura. O valor atribuído a essa lista é um dicionário, com as seguintes propriedades:

*   `type='buttons'`: define que o menu será composto por botões.
*   `showactive=False`: define que nenhum botão estará ativo inicialmente.
*   `buttons=[dict(label='Play', method='animate', args=[None, {'frame': {'duration': 100, 'redraw': True}, 'fromcurrent': True}])]`: define o botão que será exibido no menu. Este botão tem a etiqueta "Play", que é exibida no próprio botão. O método `animate` é usado para ativar a animação dos dados. O argumento `args` é uma lista que contém dois elementos: o primeiro elemento é `None`, indicando que nenhum trace (ou camada) do gráfico será afetado pela animação, e o segundo elemento é um dicionário que especifica os parâmetros da animação. O parâmetro `frame` define a duração de cada quadro da animação e a atualização de cada quadro. O parâmetro `fromcurrent` define se o quadro atual deve ser mantido ou se a animação deve ser iniciada do primeiro quadro.

```python
    updatemenus=[dict(
        type='buttons',
        showactive=False,
        buttons=[dict(
            label='Play',
            method='animate',
            args=[None, {'frame': {'duration': 100, 'redraw': True}, 'fromcurrent': True}]
        )]
    )],
```

### 7 - Dimensões do gráfico
A largura e a altura do gráfico são definidas com os parâmetros `width` e `height`, respectivamente:

```python
    width=1000,
    height=500
)
```

### 8 - Definição dos quadros (frames)
Em seguida, as configurações de animação são definidas. A variável `frames` é uma lista de objetos `Frame` do Plotly, que contém as informações dos dados para cada quadro da animação. Cada quadro é representado por um objeto `Frame` que contém um único trace, que é um objeto `Scatter`. No caso deste código, cada `Scatter` representa um ponto no gráfico, onde X é o ano e Y é o número de imigrantes.

O loop `for` é usado para criar um objeto `Frame` para cada ano no conjunto de dados, até o último ano. O método `iloc` é usado para selecionar os valores do DataFrame `dados_brasil` a partir do índice 0 até o índice atual `i+1`, o que significa que cada quadro da animação adiciona um ponto adicional ao gráfico. A cada iteração, a lista `frames` é preenchida com um novo objeto `Frame`.

Por último, a lista `frames` é atribuída à propriedade `frames` do objeto Figure (`fig`) criado anteriormente. Isso permite que a animação seja exibida no gráfico quando o botão de controle de animação é pressionado. Cada frame contém os dados do gráfico para um determinado ano e é exibido em sequência quando o botão "Play" é clicado.

```python
frames = [go.Frame(data=[go.Scatter(x=dados_brasil['ano'].iloc[:i+1], y=dados_brasil['imigrantes'].iloc[:i+1])]) for i in range(len(dados_brasil))]
fig.frames = frames
```

### 9 - Exibição do gráfico
Por fim, a função `fig.show()` é chamada para exibir o gráfico animado no notebook. Quando o botão "Play" é pressionado, o gráfico será animado, mostrando a imigração do Brasil para o Canadá no período de 1980 a 2013.

Temos o código completo abaixo:

```python
import plotly.graph_objs as go

# Criando uma figura
fig = go.Figure()

# Adicionando a linha do gráfico e definindo a espessura da linha
fig.add_trace(
    go.Scatter(x=[dados_brasil['ano'].iloc[0]], y=[dados_brasil['imigrantes'].iloc[0]], mode='lines', name='Imigrantes', line=dict(width=4))
)

# Definindo as configurações de layout
fig.update_layout(
    title=dict(
        text='<b>Imigração do Brasil para o Canadá no período de 1980 a 2013</b>',
        x=0.12,
        xanchor='left',
        font=dict(size=20)
    ),
    xaxis=dict(range=[1980, 2013], autorange=False, title='<b>Ano</b>'),
    yaxis=dict(range=[0, 3000], autorange=False, title='<b>Número de imigrantes</b>'),
    updatemenus=[dict(
        type='buttons',
        showactive=False,
        buttons=[dict(
            label='Play',
            method='animate',
            args=[None, {'frame': {'duration': 100, 'redraw': True}, 'fromcurrent': True}]
        )]
    )],
    width=1000,
    height=500
)

# Definir as configurações de animação
frames = [go.Frame(data=[go.Scatter(x=dados_brasil['ano'].iloc[:i+1], y=dados_brasil['imigrantes'].iloc[:i+1])]) for i in range(len(dados_brasil))]
fig.frames = frames

# Mostrando a figura
fig.show()
```
